```python

"""
Candidate Key Uniqueness Analysis

Purpose:
Evaluate whether the proposed candidate keys uniquely identify
records in each dataset table.

The analysis checks:
1. Null values in candidate key columns.
2. Duplicate key combinations.
3. Number of unique key combinations.

Note:
Candidate keys are assumptions based on the dataset schema.
Their validity must be confirmed through profiling.
"""

# Define the proposed candidate key for each table.
# Composite keys are used when a single column is not expected
# to uniquely identify a record.

candidate_keys = {
    "Dim_Calendar": ["Date_Key"],
    "Dim_Geography": ["Region_Key"],
    "Dim_Products": ["Product_Key"],
    "Dim_Promotions": ["Promo_Key"],
    "Dim_Salesmen": ["Salesman_Key"],
    "Dim_Stores": ["Store_Key"],
    "Fact_Sales": ["Transaction_ID"],
    "Fact_Visits": ["Visit_ID"],
    "Fact_Targets": ["Salesman_Key", "Target_Month"],
    "Fact_Market_Share": ["Date_Key", "Region_Key", "Brand_Name"]
}

key_results = []

# Iterate through each table and evaluate its proposed candidate key.
for table_name, key_columns in candidate_keys.items():

    file_path = Path(DATA_PATH) / f"{table_name}.csv"

    # Load only the candidate key columns to reduce memory usage.
    key_df = pd.read_csv(
        file_path,
        usecols=key_columns
    )

    # Count records where at least one candidate key column is missing.
    null_rows = key_df[key_columns].isnull().any(axis=1).sum()

    # Count repeated candidate key combinations.
    # keep='first' means the first occurrence is not counted as a duplicate.
    duplicate_rows = key_df.duplicated(
        subset=key_columns
    ).sum()

    # Count distinct candidate key combinations.
    unique_combinations = key_df.drop_duplicates(
        subset=key_columns
    ).shape[0]

    # Store the profiling results for the current table.
    key_results.append({
        "Table_Name": table_name,
        "Candidate_Key": " + ".join(key_columns),
        "Null_Key_Rows": null_rows,
        "Duplicate_Key_Rows": duplicate_rows,
        "Unique_Key_Combinations": unique_combinations,
        "Total_Rows": len(key_df)
    })

# Combine the results into a single DataFrame for comparison.
key_analysis_df = pd.DataFrame(key_results)

display(key_analysis_df)
```

```python

"""
Transaction ID Frequency Analysis

Purpose:
Investigate repeated Transaction_ID values in Fact_Sales
to understand whether they represent multiple product lines
within the same transaction or potential data quality issues.
"""

# Load only the transaction identifier to minimize memory usage.
sales_ids = pd.read_csv(
    Path(DATA_PATH) / "Fact_Sales.csv",
    usecols=["Transaction_ID"]
)

# Count how many records belong to each transaction.
transaction_id_counts = sales_ids["Transaction_ID"].value_counts()

# Keep only transaction IDs that appear more than once.
duplicate_transaction_ids = transaction_id_counts[
    transaction_id_counts > 1
]

# Summarize how many transactions have each repetition frequency.
frequency_distribution = (
    duplicate_transaction_ids
    .value_counts()
    .sort_index()
    .rename_axis("Rows_Per_Transaction")
    .reset_index(name="Number_of_Transactions")
)

# Display key investigation metrics.
print("Total Unique Transaction IDs:", len(transaction_id_counts))
print("Repeated Transaction IDs:", len(duplicate_transaction_ids))
print("Maximum Rows per Transaction:", transaction_id_counts.max())

# Display the frequency distribution.
display(frequency_distribution)

# Inspect the 10 most repeated transaction IDs.
display(duplicate_transaction_ids.head(10).rename("Record_Count").to_frame())
```

```python

"""
Duplicate Transaction Inspection

Purpose:
Inspect complete sales records for selected repeated Transaction_IDs
to determine whether repeated IDs represent multiple product lines
or potentially duplicated records.

The inspection uses chunk-based reading to limit memory usage.
"""

# Select a small sample of repeated transaction IDs for investigation.
sample_transaction_ids = duplicate_transaction_ids.index[:5]

# Define the columns needed to inspect transaction details.
inspection_columns = [
    "Transaction_ID",
    "Date_Key",
    "Store_Key",
    "Product_Key",
    "Salesman_Key",
    "Promo_Key",
    "Order_Source",
    "Quantity_Cases",
    "Qty_Returned",
    "Gross_Sales",
    "Discount_Amount",
    "Net_Sales",
    "COGS"
]

# Read the sales file in chunks and retain only the selected transactions.
sales_sample_chunks = []

for chunk in pd.read_csv(
    Path(DATA_PATH) / "Fact_Sales.csv",
    usecols=inspection_columns,
    chunksize=200_000
):
    matching_rows = chunk[
        chunk["Transaction_ID"].isin(sample_transaction_ids)
    ]

    if not matching_rows.empty:
        sales_sample_chunks.append(matching_rows)

# Combine the matching records into one DataFrame.
duplicate_transactions_sample = pd.concat(
    sales_sample_chunks,
    ignore_index=True
)

# Sort records to make transaction-level comparisons easier.
duplicate_transactions_sample = duplicate_transactions_sample.sort_values(
    by=["Transaction_ID", "Product_Key"]
).reset_index(drop=True)

# Display the selected transactions.
display(duplicate_transactions_sample)
```

```python

"""
Exact Duplicate Records Validation

Purpose:
Determine whether repeated Transaction_ID values represent
exact duplicate records or transactions containing different data.

The analysis:
1. Loads all columns for repeated transactions only.
2. Compares records across all columns.
3. Identifies transaction IDs with identical or different records.

This helps distinguish duplicate records from legitimate
multiple-line transactions.
"""

# Identify all columns in the sales dataset.
sales_file_path = Path(DATA_PATH) / "Fact_Sales.csv"

sales_columns = pd.read_csv(
    sales_file_path,
    nrows=0
).columns.tolist()

# Extract all records belonging to repeated transaction IDs.
repeated_sales_chunks = []

for chunk in pd.read_csv(
    sales_file_path,
    chunksize=200_000
):
    matching_rows = chunk[
        chunk["Transaction_ID"].isin(duplicate_transaction_ids.index)
    ]

    if not matching_rows.empty:
        repeated_sales_chunks.append(matching_rows)

# Combine the matching records into one DataFrame.
repeated_sales_df = pd.concat(
    repeated_sales_chunks,
    ignore_index=True
)

# Compare all columns except Transaction_ID within each transaction.
# If every column has only one distinct value, the records are identical.
comparison_columns = [
    column for column in sales_columns
    if column != "Transaction_ID"
]

variation_by_transaction = (
    repeated_sales_df
    .groupby("Transaction_ID")[comparison_columns]
    .nunique(dropna=False)
)

# Identify transactions whose records are identical across all columns.
identical_transaction_mask = variation_by_transaction.eq(1).all(axis=1)

identical_transaction_ids = variation_by_transaction.index[
    identical_transaction_mask
]

# Identify transactions containing differences between their records.
non_identical_transaction_ids = variation_by_transaction.index[
    ~identical_transaction_mask
]

# Count exact duplicate rows across the repeated transactions.
exact_duplicate_rows = repeated_sales_df.duplicated(
    subset=sales_columns
).sum()

# Summarize the validation results.
duplicate_validation_summary = pd.DataFrame({
    "Metric": [
        "Repeated Transaction IDs",
        "Identical Transaction IDs",
        "Non-Identical Transaction IDs",
        "Exact Duplicate Rows"
    ],
    "Value": [
        len(variation_by_transaction),
        len(identical_transaction_ids),
        len(non_identical_transaction_ids),
        exact_duplicate_rows
    ]
})

display(duplicate_validation_summary)

# Inspect transactions where at least one column differs.
if len(non_identical_transaction_ids) > 0:

    non_identical_sample = repeated_sales_df[
        repeated_sales_df["Transaction_ID"].isin(
            non_identical_transaction_ids[:5]
        )
    ].sort_values("Transaction_ID")

    display(non_identical_sample)

else:
    print("All repeated transactions contain identical records.")
```

```python

"""
Foreign Key Referential Integrity Analysis

Purpose:
Validate whether foreign key values in fact tables
exist in their corresponding dimension tables.

The analysis measures:
1. Total records evaluated.
2. Null foreign key records.
3. Unmatched foreign key records.
4. Distinct unmatched key values.

Implementation:
- Load only the referenced dimension keys.
- Process large fact tables using chunk-based reading.
- Scan each fact table once, even when it contains
  multiple foreign key relationships.
- Preserve the original raw data without modifications.
"""

# Define the proposed foreign key relationships.
foreign_key_relationships = {
    "Fact_Sales": {
        "Date_Key": ("Dim_Calendar", "Date_Key"),
        "Store_Key": ("Dim_Stores", "Store_Key"),
        "Product_Key": ("Dim_Products", "Product_Key"),
        "Salesman_Key": ("Dim_Salesmen", "Salesman_Key"),
        "Promo_Key": ("Dim_Promotions", "Promo_Key")
    },

    "Fact_Visits": {
        "Date_Key": ("Dim_Calendar", "Date_Key"),
        "Store_Key": ("Dim_Stores", "Store_Key"),
        "Salesman_Key": ("Dim_Salesmen", "Salesman_Key")
    },

    "Fact_Targets": {
        "Salesman_Key": ("Dim_Salesmen", "Salesman_Key"),
        "Target_Month": ("Dim_Calendar", "Date_Key")
    },

    "Fact_Market_Share": {
        "Date_Key": ("Dim_Calendar", "Date_Key"),
        "Region_Key": ("Dim_Geography", "Region_Key")
    }
}


# Load distinct, non-null keys from each referenced dimension.
# Sets allow efficient membership checks during fact table profiling.

dimension_keys = {}

for relationships in foreign_key_relationships.values():

    for dimension_table, dimension_column in relationships.values():

        dimension_key_name = (dimension_table, dimension_column)

        if dimension_key_name not in dimension_keys:

            dimension_path = Path(DATA_PATH) / f"{dimension_table}.csv"

            dimension_df = pd.read_csv(
                dimension_path,
                usecols=[dimension_column]
            )

            dimension_keys[dimension_key_name] = set(
                dimension_df[dimension_column].dropna().unique()
            )


# Initialize the results collection.
referential_integrity_results = []


# Process each fact table separately.
for fact_table, relationships in foreign_key_relationships.items():

    fact_path = Path(DATA_PATH) / f"{fact_table}.csv"

    # Read only the foreign key columns required for validation.
    foreign_key_columns = list(relationships.keys())

    # Initialize metrics for each relationship.
    relationship_metrics = {
        foreign_key: {
            "Total_Rows": 0,
            "Null_FK_Rows": 0,
            "Unmatched_FK_Rows": 0,
            "Unmatched_Keys": set()
        }
        for foreign_key in foreign_key_columns
    }

    # Process the fact table in chunks to manage memory usage.
    for chunk in pd.read_csv(
        fact_path,
        usecols=foreign_key_columns,
        chunksize=200_000
    ):

        # Validate each foreign key against its referenced dimension.
        for foreign_key, (dimension_table, dimension_column) in relationships.items():

            values = chunk[foreign_key]

            valid_keys = dimension_keys[
                (dimension_table, dimension_column)
            ]

            metrics = relationship_metrics[foreign_key]

            # Count all records evaluated for this relationship.
            metrics["Total_Rows"] += len(values)

            # Identify missing foreign key values.
            null_mask = values.isna()

            metrics["Null_FK_Rows"] += null_mask.sum()

            # Check only non-null values against the dimension keys.
            unmatched_mask = (
                ~null_mask
                & ~values.isin(valid_keys)
            )

            metrics["Unmatched_FK_Rows"] += unmatched_mask.sum()

            # Keep distinct unmatched values for further investigation.
            metrics["Unmatched_Keys"].update(
                values[unmatched_mask].unique().tolist()
            )

    # Store the final metrics for each relationship.
    for foreign_key, (dimension_table, dimension_column) in relationships.items():

        metrics = relationship_metrics[foreign_key]

        referential_integrity_results.append({
            "Fact_Table": fact_table,
            "Foreign_Key": foreign_key,
            "Dimension_Table": dimension_table,
            "Referenced_Key": dimension_column,
            "Total_Rows": metrics["Total_Rows"],
            "Null_FK_Rows": metrics["Null_FK_Rows"],
            "Unmatched_FK_Rows": metrics["Unmatched_FK_Rows"],
            "Distinct_Unmatched_Keys": len(metrics["Unmatched_Keys"]),
            "Unmatched_Key_Examples": list(
                metrics["Unmatched_Keys"]
            )[:10]
        })


# Convert the validation results into a DataFrame.
referential_integrity_df = pd.DataFrame(
    referential_integrity_results
)

# Display the validation results.
display(referential_integrity_df)
```

```python

"""
Date Format Investigation

Purpose:
Investigate unmatched Date_Key values between Fact_Sales
and Dim_Calendar.

The objective is to determine whether the mismatches
are caused by inconsistent date representations rather
than genuinely missing calendar dates.

The raw values are preserved as strings during this inspection.
"""

# Define the source file paths.
sales_file_path = Path(DATA_PATH) / "Fact_Sales.csv"
calendar_file_path = Path(DATA_PATH) / "Dim_Calendar.csv"


# Classify date strings based on their expected representation.
def classify_date_format(value):

    if pd.isna(value):
        return "Null"

    value = str(value).strip()

    if pd.Series([value]).str.fullmatch(r"\d{4}-\d{2}-\d{2}").iloc[0]:
        return "YYYY-MM-DD"

    if pd.Series([value]).str.fullmatch(r"\d{2}/\d{2}/\d{4}").iloc[0]:
        return "DD/MM/YYYY"

    return "Other / Unexpected Format"


# Inspect Date_Key values in Fact_Sales using chunk-based reading.
sales_date_format_counts = {}
sales_date_examples = {}

for chunk in pd.read_csv(
    sales_file_path,
    usecols=["Date_Key"],
    dtype={"Date_Key": "string"},
    chunksize=200_000
):

    formats = chunk["Date_Key"].apply(classify_date_format)

    # Accumulate the number of records for each date format.
    counts = formats.value_counts()

    for date_format, count in counts.items():
        sales_date_format_counts[date_format] = (
            sales_date_format_counts.get(date_format, 0) + count
        )

    # Preserve a small sample of values for each format.
    for date_format in formats.unique():

        if date_format not in sales_date_examples:
            sales_date_examples[date_format] = []

        examples = chunk.loc[
            formats == date_format, "Date_Key"
        ].dropna().unique().tolist()

        remaining_slots = 5 - len(sales_date_examples[date_format])

        sales_date_examples[date_format].extend(
            examples[:remaining_slots]
        )


# Inspect Date_Key values in Dim_Calendar.
calendar_dates = pd.read_csv(
    calendar_file_path,
    usecols=["Date_Key"],
    dtype={"Date_Key": "string"}
)

calendar_format_counts = (
    calendar_dates["Date_Key"]
    .apply(classify_date_format)
    .value_counts()
)

# Combine the format counts into a comparison table.
date_format_summary = pd.DataFrame({
    "Fact_Sales": pd.Series(sales_date_format_counts),
    "Dim_Calendar": calendar_format_counts
}).fillna(0).astype(int)

display(date_format_summary)

# Display representative raw date values from both tables.
print("Fact_Sales Date Examples:")

for date_format, examples in sales_date_examples.items():
    print(f"{date_format}: {examples}")

print("\nDim_Calendar Date Examples:")
print(calendar_dates["Date_Key"].head(10).tolist())
```

```python

"""
Date Normalization & Referential Integrity Revalidation

Purpose:
Determine whether the 60,232 previously unmatched Fact_Sales
dates are genuinely missing from Dim_Calendar or simply
stored using a different date format.

Approach:
1. Load valid dates from Dim_Calendar.
2. Read Fact_Sales in chunks.
3. Parse each date using its detected format.
4. Convert dates into a consistent YYYY-MM-DD representation.
5. Compare normalized dates against Dim_Calendar.
6. Report any remaining unmatched dates.

The original CSV files are not modified.
"""

# Load the valid calendar dates into a set for efficient lookup.
calendar_dates = pd.read_csv(
    Path(DATA_PATH) / "Dim_Calendar.csv",
    usecols=["Date_Key"],
    dtype={"Date_Key": "string"}
)

valid_calendar_dates = set(
    calendar_dates["Date_Key"].str.strip().dropna()
)


# Initialize validation metrics.
total_rows = 0
valid_normalized_rows = 0
invalid_date_rows = 0
unmatched_date_rows = 0

unmatched_date_values = set()


# Process Fact_Sales in chunks to manage memory usage.
for chunk in pd.read_csv(
    Path(DATA_PATH) / "Fact_Sales.csv",
    usecols=["Date_Key"],
    dtype={"Date_Key": "string"},
    chunksize=200_000
):

    raw_dates = chunk["Date_Key"].str.strip()

    # Initialize a temporary Series to store normalized dates.
    normalized_dates = pd.Series(
        pd.NA,
        index=chunk.index,
        dtype="string"
    )

    # Identify dates stored in YYYY-MM-DD format.
    iso_mask = raw_dates.str.fullmatch(
        r"\d{4}-\d{2}-\d{2}"
    ).fillna(False)

    # Identify dates stored in DD/MM/YYYY format.
    day_first_mask = raw_dates.str.fullmatch(
        r"\d{2}/\d{2}/\d{4}"
    ).fillna(False)

    # Parse ISO-formatted dates.
    normalized_dates.loc[iso_mask] = pd.to_datetime(
        raw_dates.loc[iso_mask],
        format="%Y-%m-%d",
        errors="coerce"
    ).dt.strftime("%Y-%m-%d")

    # Parse day-first dates.
    normalized_dates.loc[day_first_mask] = pd.to_datetime(
        raw_dates.loc[day_first_mask],
        format="%d/%m/%Y",
        errors="coerce"
    ).dt.strftime("%Y-%m-%d")

    # Track records that could not be parsed.
    invalid_mask = normalized_dates.isna()

    invalid_date_rows += invalid_mask.sum()

    # Count successfully normalized records.
    valid_normalized_rows += (~invalid_mask).sum()

    # Identify normalized dates missing from the calendar dimension.
    unmatched_mask = (
        normalized_dates.notna()
        & ~normalized_dates.isin(valid_calendar_dates)
    )

    unmatched_date_rows += unmatched_mask.sum()

    # Collect distinct unmatched dates for further investigation.
    unmatched_date_values.update(
        normalized_dates.loc[unmatched_mask].unique().tolist()
    )

    total_rows += len(chunk)


# Summarize the validation results.
date_validation_summary = pd.DataFrame({
    "Metric": [
        "Total Fact_Sales Rows",
        "Successfully Parsed Date Rows",
        "Invalid or Missing Date Rows",
        "Unmatched Date Rows After Normalization",
        "Distinct Unmatched Dates After Normalization"
    ],
    "Value": [
        total_rows,
        valid_normalized_rows,
        invalid_date_rows,
        unmatched_date_rows,
        len(unmatched_date_values)
    ]
})

display(date_validation_summary)

# Display examples only if unmatched dates remain.
if unmatched_date_values:
    print("Unmatched Date Examples:")
    print(sorted(unmatched_date_values)[:20])

else:
    print("All successfully parsed Fact_Sales dates exist in Dim_Calendar.")
```

```python

"""
Candidate Grain Validation

Purpose:
Check whether the proposed grain columns uniquely identify
records in each fact table.

This is a profiling step only. No source data is modified.
"""

from pathlib import Path
import pandas as pd

# Define the proposed grain for each fact table.
fact_grains = {
    "Fact_Sales": ["Transaction_ID"],
    "Fact_Visits": ["Visit_ID"],
    "Fact_Targets": ["Salesman_Key", "Target_Month"],
    "Fact_Market_Share": ["Date_Key", "Region_Key", "Brand_Name"]
}

grain_results = []

for table_name, grain_columns in fact_grains.items():

    # Read only the columns needed for grain validation.
    df = pd.read_csv(
        Path(DATA_PATH) / f"{table_name}.csv",
        usecols=grain_columns,
        dtype="string"
    )

    total_rows = len(df)

    # A record with any missing grain component cannot be
    # reliably identified using the proposed grain.
    null_key_mask = df[grain_columns].isna().any(axis=1)
    null_key_rows = null_key_mask.sum()

    # Count occurrences of each proposed grain combination.
    grain_counts = (
        df.groupby(grain_columns, dropna=False)
        .size()
    )

    duplicate_groups = grain_counts[grain_counts > 1]

    grain_results.append({
        "Fact Table": table_name,
        "Candidate Grain": " + ".join(grain_columns),
        "Total Rows": total_rows,
        "Null Grain Rows": null_key_rows,
        "Unique Grain Combinations": len(grain_counts),
        "Duplicate Grain Groups": len(duplicate_groups),
        "Duplicate Excess Rows": (
            duplicate_groups.sum() - len(duplicate_groups)
        ),
        "Rows in Duplicate Groups": duplicate_groups.sum(),
        "Maximum Rows per Grain": (
            grain_counts.max() if not grain_counts.empty else 0
        )
    })

grain_validation_summary = pd.DataFrame(grain_results)

display(grain_validation_summary)
```

```python

"""
Missing Values & Data Completeness Profiling

Purpose:
Identify missing values across all dataset columns,
including null values and empty/whitespace-only strings.

Large fact tables are processed in chunks to reduce memory usage.
No source data is modified.
"""

from pathlib import Path
import pandas as pd

data_dir = Path(DATA_PATH)

csv_files = sorted(data_dir.glob("*.csv"))

missing_results = []

for file_path in csv_files:

    table_name = file_path.stem

    # Process each table in chunks to handle large fact tables safely.
    missing_counts = {}
    total_rows = 0
    rows_with_missing = 0

    for chunk in pd.read_csv(
        file_path,
        dtype="string",
        chunksize=200_000
    ):

        total_rows += len(chunk)

        # Treat nulls and empty/whitespace-only strings as missing.
        missing_mask = (
            chunk.isna()
            | chunk.apply(
                lambda column: column.str.strip().eq("").fillna(False)
            )
        )

        # Accumulate missing counts for each column.
        if not missing_counts:
            missing_counts = {
                column: 0 for column in chunk.columns
            }

        for column in chunk.columns:
            missing_counts[column] += missing_mask[column].sum()

        # Count records containing at least one missing value.
        rows_with_missing += missing_mask.any(axis=1).sum()

    # Generate a column-level completeness report.
    for column, missing_count in missing_counts.items():

        missing_results.append({
            "Table": table_name,
            "Column": column,
            "Total Rows": total_rows,
            "Missing Values": missing_count,
            "Missing %": round(
                (missing_count / total_rows) * 100, 2
            ) if total_rows else 0
        })

    print(
        f"{table_name}: {rows_with_missing:,} rows "
        f"with at least one missing value "
        f"out of {total_rows:,}"
    )


missing_values_report = pd.DataFrame(missing_results)

# Show columns with missing values first.
missing_values_report = missing_values_report.sort_values(
    by=["Missing Values", "Table"],
    ascending=[False, True]
).reset_index(drop=True)

display(missing_values_report)
```

```python

"""
Missing Value Investigation

Purpose:
1. Determine whether missing Return_Reason values are
   consistent with transactions that have no returns.
2. Inspect salesmen whose Supervisor_Name is missing.

No source data is modified.
"""

from pathlib import Path
import pandas as pd

data_dir = Path(DATA_PATH)


# --------------------------------------------------
# Part A: Return_Reason vs Qty_Returned
# --------------------------------------------------

return_summary = {
    "No Return": {"Rows": 0, "Missing Reason": 0, "Reason Present": 0},
    "Positive Return": {"Rows": 0, "Missing Reason": 0, "Reason Present": 0},
    "Negative Return": {"Rows": 0, "Missing Reason": 0, "Reason Present": 0}
}

positive_return_without_reason = []
zero_return_with_reason = []

for chunk in pd.read_csv(
    data_dir / "Fact_Sales.csv",
    usecols=["Qty_Returned", "Return_Reason"],
    chunksize=200_000
):

    # Convert quantities to numeric for reliable comparisons.
    qty = pd.to_numeric(chunk["Qty_Returned"], errors="coerce")

    reason = chunk["Return_Reason"].str.strip()

    reason_missing = (
        reason.isna() | reason.eq("").fillna(False)
    )

    categories = {
        "No Return": qty == 0,
        "Positive Return": qty > 0,
        "Negative Return": qty < 0
    }

    for category, mask in categories.items():

        return_summary[category]["Rows"] += mask.sum()

        return_summary[category]["Missing Reason"] += (
            (mask & reason_missing).sum()
        )

        return_summary[category]["Reason Present"] += (
            (mask & ~reason_missing).sum()
        )

    # Collect examples of potential business-rule violations.
    positive_missing = (qty > 0) & reason_missing
    zero_present = (qty == 0) & ~reason_missing

    if positive_missing.any() and len(positive_return_without_reason) < 10:
        positive_return_without_reason.extend(
            chunk.loc[
                positive_missing,
                ["Qty_Returned", "Return_Reason"]
            ].to_dict("records")
        )

    if zero_present.any() and len(zero_return_with_reason) < 10:
        zero_return_with_reason.extend(
            chunk.loc[
                zero_present,
                ["Qty_Returned", "Return_Reason"]
            ].to_dict("records")
        )


return_reason_report = pd.DataFrame.from_dict(
    return_summary,
    orient="index"
).reset_index(names="Return Status")

display(return_reason_report)

print("Examples: Positive Returns Without a Reason")
display(pd.DataFrame(positive_return_without_reason))

print("Examples: Zero Returns With a Reason")
display(pd.DataFrame(zero_return_with_reason))


# --------------------------------------------------
# Part B: Missing Supervisor_Name
# --------------------------------------------------

salesmen = pd.read_csv(
    data_dir / "Dim_Salesmen.csv",
    dtype="string"
)

missing_supervisor = salesmen[
    salesmen["Supervisor_Name"].isna()
    | salesmen["Supervisor_Name"].str.strip().eq("").fillna(False)
]

print("Salesmen With Missing Supervisor_Name:")
display(missing_supervisor)

print("Supervisor Name Distribution:")
display(
    salesmen["Supervisor_Name"]
    .fillna("<MISSING>")
    .value_counts()
    .rename_axis("Supervisor_Name")
    .reset_index(name="Salesmen Count")
)
```

```python

"""
Supervisor Missingness Investigation

Purpose:
Explore whether missing Supervisor_Name values follow
a systematic pattern across employee keys and positions.

This is an investigation only; no values are imputed.
"""

salesmen = pd.read_csv(
    Path(DATA_PATH) / "Dim_Salesmen.csv"
)

# Identify records with missing or blank supervisor names.
missing_mask = (
    salesmen["Supervisor_Name"].isna()
    | salesmen["Supervisor_Name"].astype("string").str.strip().eq("").fillna(False)
)

missing = salesmen[missing_mask].copy()
complete = salesmen[~missing_mask].copy()


# 1. Check whether missing employee keys follow a pattern.
missing["Salesman_Key_Mod_15"] = (
    pd.to_numeric(missing["Salesman_Key"]) % 15
)

print("Missing Salesman Keys:")
display(
    missing[
        ["Salesman_Key", "Salesman_Name", "Position",
         "Salesman_Key_Mod_15"]
    ]
)


# 2. Compare position distributions.
position_comparison = pd.concat(
    [
        missing["Position"].value_counts().rename("Missing Supervisor"),
        complete["Position"].value_counts().rename("Supervisor Present")
    ],
    axis=1
).fillna(0).astype(int)

position_comparison["Missing % Within Position"] = (
    position_comparison["Missing Supervisor"]
    / (
        position_comparison["Missing Supervisor"]
        + position_comparison["Supervisor Present"]
    )
    * 100
).round(2)

print("Position Distribution Comparison:")
display(position_comparison)


# 3. Compare basic employee characteristics.
print("Missing Supervisor Employees:")
display(
    missing[
        ["Position", "Commission_Rate", "Hiring_Date", "Sales_Tool"]
    ].describe(include="all")
)

print("Employees With Supervisor:")
display(
    complete[
        ["Position", "Commission_Rate", "Hiring_Date", "Sales_Tool"]
    ].describe(include="all")
)
```

```python

"""
Date Coverage & Format Profiling

Purpose:
Understand the date coverage and raw date representations
across all fact tables and the calendar dimension.

This is a profiling step. No dates are modified.
"""

from pathlib import Path
import pandas as pd

data_dir = Path(DATA_PATH)

date_columns = {
    "Dim_Calendar": ["Date_Key"],
    "Fact_Sales": ["Date_Key"],
    "Fact_Visits": ["Date_Key"],
    "Fact_Targets": ["Target_Month"],
    "Fact_Market_Share": ["Date_Key"]
}

date_profiles = []

for table_name, columns in date_columns.items():

    for column in columns:

        total_rows = 0
        format_counts = {}
        examples = {}

        min_date = None
        max_date = None
        unique_values = set()

        for chunk in pd.read_csv(
            data_dir / f"{table_name}.csv",
            usecols=[column],
            dtype="string",
            chunksize=200_000
        ):

            values = chunk[column].str.strip().dropna()
            total_rows += len(chunk)

            # Identify common date representations without
            # assuming that every field has daily granularity.
            patterns = {
                "YYYY-MM-DD": r"\d{4}-\d{2}-\d{2}",
                "DD/MM/YYYY": r"\d{2}/\d{2}/\d{4}",
                "YYYY-MM": r"\d{4}-\d{2}",
                "Other": None
            }

            classified = pd.Series(
                "Other",
                index=values.index,
                dtype="string"
            )

            for format_name, pattern in patterns.items():
                if pattern is not None:
                    mask = values.str.fullmatch(pattern).fillna(False)
                    classified.loc[mask] = format_name

            for format_name, count in classified.value_counts().items():
                format_counts[format_name] = (
                    format_counts.get(format_name, 0) + count
                )

            for format_name in patterns:
                format_values = values[classified == format_name]

                if len(format_values) and format_name not in examples:
                    examples[format_name] = (
                        format_values.head(3).tolist()
                    )

            unique_values.update(values.unique().tolist())

        date_profiles.append({
            "Table": table_name,
            "Column": column,
            "Total Rows": total_rows,
            "Distinct Values": len(unique_values),
            "Format Distribution": format_counts,
            "Format Examples": examples
        })

date_coverage_report = pd.DataFrame(date_profiles)

display(date_coverage_report)
```

```python

"""
Normalized Date Coverage Validation

Purpose:
Validate actual date coverage after normalizing different
date formats into a common YYYY-MM-DD representation.

The analysis distinguishes daily facts from monthly facts.
No source data is modified.
"""

from pathlib import Path
import pandas as pd

data_dir = Path(DATA_PATH)

date_columns = {
    "Dim_Calendar": "Date_Key",
    "Fact_Sales": "Date_Key",
    "Fact_Visits": "Date_Key",
    "Fact_Targets": "Target_Month",
    "Fact_Market_Share": "Date_Key"
}

# Load the calendar's canonical dates for reference.
calendar_raw = pd.read_csv(
    data_dir / "Dim_Calendar.csv",
    usecols=["Date_Key"],
    dtype="string"
)

calendar_dates = pd.to_datetime(
    calendar_raw["Date_Key"],
    format="%Y-%m-%d",
    errors="coerce"
).dropna()

valid_calendar_dates = set(calendar_dates.dt.strftime("%Y-%m-%d"))


# Normalize supported date formats without changing source files.
def normalize_dates(values):

    normalized = pd.Series(
        pd.NaT,
        index=values.index,
        dtype="datetime64[ns]"
    )

    iso_mask = values.str.fullmatch(
        r"\d{4}-\d{2}-\d{2}"
    ).fillna(False)

    day_first_mask = values.str.fullmatch(
        r"\d{2}/\d{2}/\d{4}"
    ).fillna(False)

    normalized.loc[iso_mask] = pd.to_datetime(
        values.loc[iso_mask],
        format="%Y-%m-%d",
        errors="coerce"
    )

    normalized.loc[day_first_mask] = pd.to_datetime(
        values.loc[day_first_mask],
        format="%d/%m/%Y",
        errors="coerce"
    )

    return normalized


coverage_results = []

for table_name, column in date_columns.items():

    total_rows = 0
    invalid_rows = 0
    normalized_dates_set = set()

    for chunk in pd.read_csv(
        data_dir / f"{table_name}.csv",
        usecols=[column],
        dtype="string",
        chunksize=200_000
    ):

        total_rows += len(chunk)

        raw_values = chunk[column].str.strip()
        parsed = normalize_dates(raw_values)

        invalid_rows += parsed.isna().sum()

        normalized_dates_set.update(
            parsed.dropna().dt.strftime("%Y-%m-%d").tolist()
        )

    # Convert canonical values back to datetime for range analysis.
    unique_dates = pd.to_datetime(
        sorted(normalized_dates_set),
        format="%Y-%m-%d"
    )

    # Identify dates not represented in the calendar dimension.
    unmatched_dates = (
        normalized_dates_set - valid_calendar_dates
    )

    # Monthly coverage helps distinguish monthly snapshots
    # from daily activity.
    unique_months = unique_dates.to_period("M").nunique()

    non_month_start_dates = sum(
        date.day != 1 for date in unique_dates
    )

    coverage_results.append({
        "Table": table_name,
        "Date Column": column,
        "Total Rows": total_rows,
        "Invalid Date Rows": invalid_rows,
        "Distinct Normalized Dates": len(unique_dates),
        "Distinct Months": unique_months,
        "Minimum Date": (
            unique_dates.min().strftime("%Y-%m-%d")
            if len(unique_dates) else None
        ),
        "Maximum Date": (
            unique_dates.max().strftime("%Y-%m-%d")
            if len(unique_dates) else None
        ),
        "Dates Outside Calendar": len(unmatched_dates),
        "Non-Month-Start Dates": non_month_start_dates
    })


date_coverage_validation = pd.DataFrame(coverage_results)

display(date_coverage_validation)
```

```python

"""
Daily and Monthly Date Coverage Consistency

Purpose:
Compare daily fact-table date coverage, identify missing
calendar days, and validate continuity of monthly records.

No source data is modified.
"""

from pathlib import Path
import pandas as pd

data_dir = Path(DATA_PATH)


def load_normalized_date_set(table_name, column_name):
    """Load dates in chunks and normalize supported date formats."""

    date_set = set()

    for chunk in pd.read_csv(
        data_dir / f"{table_name}.csv",
        usecols=[column_name],
        dtype="string",
        chunksize=200_000
    ):

        values = chunk[column_name].str.strip()

        iso_mask = values.str.fullmatch(
            r"\d{4}-\d{2}-\d{2}"
        ).fillna(False)

        day_first_mask = values.str.fullmatch(
            r"\d{2}/\d{2}/\d{4}"
        ).fillna(False)

        parsed = pd.Series(
            pd.NaT,
            index=values.index,
            dtype="datetime64[ns]"
        )

        parsed.loc[iso_mask] = pd.to_datetime(
            values.loc[iso_mask],
            format="%Y-%m-%d",
            errors="coerce"
        )

        parsed.loc[day_first_mask] = pd.to_datetime(
            values.loc[day_first_mask],
            format="%d/%m/%Y",
            errors="coerce"
        )

        date_set.update(
            parsed.dropna().dt.normalize().tolist()
        )

    return date_set


# Load normalized daily date coverage.
sales_dates = load_normalized_date_set(
    "Fact_Sales", "Date_Key"
)

visits_dates = load_normalized_date_set(
    "Fact_Visits", "Date_Key"
)

# Compare the two daily fact tables.
sales_only_dates = sales_dates - visits_dates
visits_only_dates = visits_dates - sales_dates

# Identify gaps within the observed date range.
observed_min = min(sales_dates | visits_dates)
observed_max = max(sales_dates | visits_dates)

expected_daily_dates = set(
    pd.date_range(observed_min, observed_max, freq="D")
)

missing_daily_dates = (
    expected_daily_dates - (sales_dates | visits_dates)
)

daily_comparison = pd.DataFrame([
    {
        "Metric": "Fact_Sales Distinct Dates",
        "Value": len(sales_dates)
    },
    {
        "Metric": "Fact_Visits Distinct Dates",
        "Value": len(visits_dates)
    },
    {
        "Metric": "Dates in Sales Only",
        "Value": len(sales_only_dates)
    },
    {
        "Metric": "Dates in Visits Only",
        "Value": len(visits_only_dates)
    },
    {
        "Metric": "Missing Calendar Days in Observed Range",
        "Value": len(missing_daily_dates)
    }
])

print("Daily Date Coverage Comparison:")
display(daily_comparison)

print("Examples of Missing Daily Dates:")
print(sorted(missing_daily_dates)[:30])


# Validate continuity of monthly fact-table coverage.
monthly_tables = {
    "Fact_Targets": "Target_Month",
    "Fact_Market_Share": "Date_Key"
}

monthly_results = []

for table_name, column_name in monthly_tables.items():

    dates = load_normalized_date_set(
        table_name, column_name
    )

    observed_months = {
        date.to_period("M") for date in dates
    }

    expected_months = set(
        pd.period_range(
            min(observed_months),
            max(observed_months),
            freq="M"
        )
    )

    missing_months = expected_months - observed_months

    monthly_results.append({
        "Table": table_name,
        "Observed Months": len(observed_months),
        "Expected Months": len(expected_months),
        "Missing Months": len(missing_months),
        "Missing Month Examples": [
            str(month) for month in sorted(missing_months)[:10]
        ]
    })

monthly_continuity_report = pd.DataFrame(monthly_results)

print("Monthly Coverage Continuity:")
display(monthly_continuity_report)
```
